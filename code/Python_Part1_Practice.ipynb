## 5. Conditions and Loops

```python
score = 0.83
if score >= 0.90:
    category = "excellent"
elif score >= 0.75:
    category = "good"
else:
    category = "needs review"
print(category)

for index, feature in enumerate(features):
    print(index, feature)
```
