# PART 01 — Python Fundamentals
## 1.1 What is Python?

Python is a general-purpose programming language known for readable syntax. It is used in automation, data analysis, machine learning, web development, and many other areas.

**Key idea:** a program is a set of instructions that the computer executes.

```python
print("Hello, HBE_2206!")
```

## 1.2 Your First Program

`print()` displays information in the output area. Text values are written inside quotation marks.

```python
print("My AI Learning Journey")
print("Python Part 1")
print(2206)
```

## 1.3 Comments

Comments explain code to people and are ignored by Python when the program runs. A single-line comment begins with `#`.

```python
# This is a comment
print("Comments make code easier to understand.")
```

## 1.4 Variables

A variable is a name that refers to a value. Use meaningful names that describe the data.

```python
learner = "HBE_2206"
lessons_completed = 3
is_learning_ai = True

print(learner)
print(lessons_completed)
print(is_learning_ai)
```

## 1.5 Variable Naming Rules

- Names can contain letters, digits, and underscores.
- A name cannot start with a digit.
- Names are case-sensitive (`score` and `Score` differ).
- Avoid Python keywords and prefer descriptive `snake_case` names.

```python
student_name = "Alex"
quiz_score = 95
print(student_name, quiz_score)
```

# PART 02 — Data Types & Type Conversion
## 2.1 Integers and Floats

`int` represents whole numbers. `float` represents numbers with a decimal part.

```python
age = 19
accuracy = 0.92
print(type(age))
print(type(accuracy))
```

## 2.2 Strings

A string is text surrounded by quotes. Strings can contain names, messages, labels, and other textual data.

```python
topic = "Machine Learning"
print(type(topic))
print(topic)
```

## 2.3 Booleans

A Boolean value is either `True` or `False`. Booleans often represent a condition or state.

```python
is_ready = True
has_error = False
print(is_ready, has_error)
print(type(is_ready))
```

## 2.4 Checking Types

Use `type(value)` to inspect a value's type. This is useful when debugging unexpected results.

```python
values = ["Python", 2206, 3.14, False]
for value in values:
    print(value, "->", type(value))
```

## 2.5 Type Conversion

Convert compatible values with `int()`, `float()`, and `str()`. Conversion can fail if the text does not represent a valid number.

```python
age_text = "20"
age_number = int(age_text)
price = float("12.5")
print(age_number + 1)
print(price)
```

# PART 03 — Operators & Expressions
## 3.1 Arithmetic Operators

Common operators: `+`, `-`, `*`, `/`, `//`, `%`, and `**`.

```python
a, b = 17, 5
print(a + b)
print(a / b)
print(a // b)
print(a % b)
print(2 ** 3)
```

## 3.2 Comparison Operators

Comparisons return Boolean values: `==`, `!=`, `>`, `<`, `>=`, `<=`.

```python
score = 85
print(score >= 60)
print(score == 100)
print(score != 0)
```

## 3.3 Logical Operators

`and` requires both conditions to be true, `or` requires at least one, and `not` reverses a Boolean result.

```python
age = 20
has_id = True
print(age >= 18 and has_id)
print(age < 18 or has_id)
print(not has_id)
```

## 3.4 Assignment Operators

Assignment updates a variable. Compound forms such as `+=` combine an operation with assignment.

```python
count = 2
count += 3
count *= 2
print(count)
```

## 3.5 Membership and Identity

`in` checks membership. `is` checks whether two references point to the same object; it is not a replacement for value equality (`==`).

```python
tools = ["Python", "Git", "Jupyter"]
print("Python" in tools)
first = None
print(first is None)
```

# PART 04 — Input, Output & Strings
## 4.1 Output with print()

`print()` can display multiple values. Use `sep` and `end` to control formatting.

```python
print("HBE_2206", "Python", sep=" | ")
print("Learning", end=" ")
print("in public")
```

## 4.2 User Input

`input()` reads text and returns a string. Convert the value before performing numeric calculations.

```python
# Uncomment these lines when running interactively:
# name = input("Your name: ")
# print(f"Welcome, {name}!")
raw_number = "42"
number = int(raw_number)
print(number + 8)
```

## 4.3 String Indexing

Indexing gets one character. Positive indexes start at zero; negative indexes count from the end.

```python
language = "Python"
print(language[0])
print(language[-1])
```

## 4.4 String Slicing

Slicing uses `text[start:stop:step]`. The stop index is excluded.

```python
language = "Python"
print(language[0:3])
print(language[2:])
print(language[::-1])
```

## 4.5 Common String Methods

Methods such as `.lower()`, `.upper()`, `.strip()`, `.replace()`, and `.split()` return useful transformed values.

```python
message = "  Learn Python  "
print(message.strip())
print(message.lower())
print(message.replace("Python", "AI"))
print("red,blue,green".split(","))
```

## 4.6 f-Strings

Use an f-string to insert expressions into text. Prefix the string with `f` and place expressions inside `{}`.

```python
name = "HBE_2206"
completed = 4
print(f"{name} completed {completed} lessons.")
```

# PART 05 — Python Data Structures
## 5.1 Lists

Lists are ordered and mutable. They can store multiple values and allow duplicates.

```python
topics = ["Python", "Data", "AI"]
topics.append("Robotics")
print(topics)
print(topics[0])
```

## 5.2 Tuples

Tuples are ordered collections that cannot be changed after creation. They are useful for fixed groups of values.

```python
image_size = (224, 224, 3)
print(image_size[0])
print(len(image_size))
```

## 5.3 Sets

Sets store unique elements. They are useful for removing duplicates and testing membership. Their display order should not be relied upon.

```python
labels = {"cat", "dog", "cat"}
print(len(labels))
print("dog" in labels)
```

## 5.4 Dictionaries

Dictionaries store key-value pairs. Access values by key and use `.get()` when a key may be absent.

```python
metrics = {"accuracy": 0.92, "recall": 0.88}
print(metrics["accuracy"])
print(metrics.get("precision", 0.0))
```

## 5.5 Mutability: List vs Tuple

A list can be modified in place. A tuple cannot have its elements reassigned.

```python
scores = [80, 90]
scores[0] = 85
print(scores)
fixed_point = (2, 4)
print(fixed_point)
```

## 5.6 Nested Structures

Collections can contain other collections. Keep nesting readable and access each level deliberately.

```python
student = {"name": "Sam", "scores": [88, 94, 91]}
print(student["name"])
print(student["scores"][1])
```

# PART 06 — Conditions & Decision Making
## 6.1 The if Statement

Run a block only when its condition is true. Indentation defines the block.

```python
temperature = 28
if temperature > 25:
    print("It is warm.")
```

## 6.2 if and else

Use `else` for the alternative path when the `if` condition is false.

```python
score = 72
if score >= 50:
    print("Pass")
else:
    print("Try again")
```

## 6.3 Multiple Branches with elif

Use `elif` to test another condition when previous conditions were false.

```python
score = 83
if score >= 90:
    result = "Excellent"
elif score >= 75:
    result = "Good"
else:
    result = "Keep practicing"
print(result)
```

## 6.4 Nested Conditions

A condition can appear inside another condition. Avoid deep nesting when a simpler structure is possible.

```python
age = 21
has_ticket = True
if age >= 18:
    if has_ticket:
        print("Entry allowed")
```

## 6.5 Conditional Expression

A short conditional expression chooses between two values.

```python
score = 67
status = "Pass" if score >= 50 else "Fail"
print(status)
```

# PART 07 — Loops & Iteration
## 7.1 for Loops

A `for` loop processes each item in an iterable, such as a list or string.

```python
for topic in ["Python", "ML", "AI"]:
    print(topic)
```

## 7.2 while Loops

A `while` loop repeats as long as its condition is true. Make sure the condition can eventually become false.

```python
counter = 1
while counter <= 3:
    print(counter)
    counter += 1
```

## 7.3 range()

`range(stop)` begins at zero and excludes the stop value. `range(start, stop, step)` controls the start and step.

```python
print(list(range(5)))
print(list(range(-5, 6, 5)))
```

## 7.4 break and continue

`break` exits the loop. `continue` skips to the next iteration.

```python
for number in range(1, 6):
    if number == 2:
        continue
    if number == 5:
        break
    print(number)
```

## 7.5 Nested Loops

A nested loop runs one loop inside another. Track each loop variable carefully.

```python
for row in range(2):
    for column in range(3):
        print(row, column)
```

## 7.6 enumerate()

`enumerate()` gives both an index and an item while looping.

```python
features = ["age", "fare", "class"]
for index, feature in enumerate(features, start=1):
    print(index, feature)
```

# PART 08 — Comprehensions & Problem Solving
## 8.1 List Comprehensions

A list comprehension builds a list from an iterable in a concise form.

```python
squares = [number ** 2 for number in range(1, 6)]
print(squares)
```

## 8.2 Filtering with a Comprehension

Add an `if` condition to keep only items that meet a requirement.

```python
scores = [0.92, 0.45, 0.87, 0.31, 0.96]
high_scores = [score for score in scores if score >= 0.80]
print(high_scores)
```

## 8.3 Common Errors

- `SyntaxError`: code structure is invalid.
- `NameError`: a name is not defined.
- `TypeError`: an operation is incompatible with a value's type.
- `IndexError`: a sequence index is out of range.

Read the last line of the traceback first, then inspect the line that caused the error.

```python
values = [10, 20, 30]
print(values[1])
# Try values[10] to inspect an IndexError, then fix the index.
```

## 8.4 Debugging Checklist

1. Read the full error message.
2. Identify the exact line.
3. Check variable names and types.
4. Print intermediate values when needed.
5. Change one thing at a time and rerun.
