## 2. Comprehensions


```python
squares = [n * n for n in range(1, 6)]
even_numbers = [n for n in range(10) if n % 2 == 0]
print(squares, even_numbers)
```

## 3. Exceptions and validation


```python
def parse_score(value):
    try:
        score = float(value)
    except ValueError:
        return None
    return score if 0 <= score <= 100 else None

print(parse_score("87.5"))
print(parse_score("not a score"))
```

## 4. Files and JSON


```python
import json
from pathlib import Path

record = {"name": "Hazem", "score": 92}
path = Path("student.json")
path.write_text(json.dumps(record, indent=2), encoding="utf-8")
loaded = json.loads(path.read_text(encoding="utf-8"))
print(loaded)
```

## 5. Object-oriented programming


```python
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def passed(self):
        return self.score >= 50

student = Student("Mona", 88)
print(student.name, student.passed())
```

## 6. Testing


```python
def multiply(a, b):
    return a * b

assert multiply(3, 4) == 12
assert multiply(-2, 5) == -10
print("Tests passed")
```
