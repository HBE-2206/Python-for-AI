## 2. What is Matplotlib?
Matplotlib is a Python library for creating static, animated, and interactive visualizations. Its `matplotlib.pyplot` interface provides plotting functions. It is commonly used in exploratory data analysis (EDA), reports, and model evaluation. A chart can reveal patterns, but it does not by itself prove causation.

## 4. Basic plotting workflow
Prepare data → call a plotting function → add title and labels → optionally add a legend/grid → display with `plt.show()`.

````python
x = [1, 2, 3, 4, 5]
y = [2, 4, 3, 7, 6]

plt.plot(x, y)
plt.title("A Simple Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
````

Expected result: a line connects five points; the horizontal axis is X values and the vertical axis is Y values.

## 5. Line chart — show trends
Use a line chart when the order matters, especially for time or sequence data.

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12, 15, 13, 18, 22, 21]

plt.plot(months, sales, marker="o", linewidth=2)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (units)")
plt.grid(True, alpha=0.3)
plt.show()
````

Expected result: overall upward trend, a dip in March, a rise through May, then a slight dip in June.

### Multiple lines and legend
Use labels and `plt.legend()` to distinguish multiple data series.

````python
months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [10, 14, 13, 18]
product_b = [8, 11, 15, 16]

plt.plot(months, product_a, marker="o", label="Product A")
plt.plot(months, product_b, marker="s", label="Product B")
plt.title("Product Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

Expected result: two labeled lines appear on the same axes.

## 6. Bar chart — compare categories
Bar charts compare discrete categories. When bar length represents magnitude, the axis should generally start at zero to avoid exaggerating differences.

````python
departments = ["AI", "Software", "Mechanical", "Electrical"]
members = [12, 18, 10, 14]

plt.bar(departments, members, color="#FF6B35")
plt.title("Team Members by Department")
plt.xlabel("Department")
plt.ylabel("Number of Members")
plt.show()
````

Expected result: Software has the tallest bar; Mechanical has the shortest.

### Horizontal bars
Useful when category labels are long.

````python
courses = ["Python Basics", "Data Analysis", "Machine Learning"]
hours = [10, 16, 24]

plt.barh(courses, hours, color="#FFDC52")
plt.title("Study Hours by Course")
plt.xlabel("Hours")
plt.ylabel("Course")
plt.show()
````

Expected result: three horizontal bars; Machine Learning is longest.

## 7. Scatter plot — relationship between numeric variables
Each point represents a pair of numeric values. Scatter plots can reveal clusters, trends, and outliers; association does not prove causation.

````python
study_hours = [1, 2, 2.5, 3, 4, 5, 6, 7]
exam_score = [48, 52, 58, 61, 68, 73, 82, 88]

plt.scatter(study_hours, exam_score, color="#2878B5", s=70)
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True, alpha=0.3)
plt.show()
````

Expected result: points generally rise from left to right, indicating a positive association in this example.

## 8. Histogram — distribution
A histogram groups numeric values into bins and shows the frequency in each bin. Bin width/count affects the shape you see.

````python
rng = np.random.default_rng(42)
scores = rng.normal(loc=70, scale=10, size=300)

plt.hist(scores, bins=12, edgecolor="white", color="#2878B5")
plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
````

Expected result: a roughly bell-shaped distribution centered near 70. A fixed random seed makes the sample reproducible.

## 9. Pie chart — parts of a whole
Use only when values are parts of the same whole and categories are mutually exclusive. With many categories, a bar chart is usually easier to compare.

````python
labels = ["Python", "SQL", "Excel", "Other"]
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Tools Used in a Survey")
plt.axis("equal")
plt.show()
````

Expected result: four slices labeled with percentages that total 100%.

## 10. Customize plots
Titles and axis labels explain the chart. Include units where relevant. Markers and line styles help distinguish series; avoid unnecessary decoration.

````python
x = [1, 2, 3, 4, 5]
y = [3, 5, 4, 8, 7]

plt.plot(x, y, color="#FF6B35", linestyle="--", marker="o",
         markersize=7, linewidth=2, label="Measured values")
plt.title("Customized Line Chart", fontsize=14)
plt.xlabel("Experiment number")
plt.ylabel("Measurement (units)")
plt.legend()
plt.grid(True, alpha=0.25)
plt.show()
````

Expected result: orange dashed line with circular markers, labels, legend, and a light grid.

## 11. Figure size and subplots
A figure is the full canvas; an Axes is one plotting area. Subplots place multiple views in one figure.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot([1, 2, 3, 4], [2, 4, 3, 5], marker="o")
axes[0].set_title("Line Chart")
axes[0].set_xlabel("Step")
axes[0].set_ylabel("Value")

axes[1].bar(["A", "B", "C"], [5, 3, 7])
axes[1].set_title("Bar Chart")
axes[1].set_ylabel("Count")

fig.suptitle("Two Views of Example Data")
fig.tight_layout()
plt.show()
````

Expected result: line chart on the left and bar chart on the right. `tight_layout()` helps prevent labels from overlapping.

## 12. Annotations
Use an annotation to call attention to a specific point. Keep annotations sparse to avoid clutter.

````python
days = [1, 2, 3, 4, 5]
temperature = [20, 22, 21, 27, 25]

plt.plot(days, temperature, marker="o")
plt.annotate("Highest value", xy=(4, 27), xytext=(2.8, 28),
             arrowprops={"arrowstyle": "->"})
plt.title("Daily Temperature")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.show()
````

Expected result: an arrow points to the highest value, 27°C on day 4.

## 13. Save a chart
Call `savefig()` before `show()` for predictable results. `dpi` controls raster resolution; `bbox_inches="tight"` helps include labels.

````python
plt.figure(figsize=(7, 4))
plt.plot([1, 2, 3, 4], [4, 6, 5, 9], marker="o")
plt.title("Saved Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.savefig("matplotlib_example.png", dpi=200, bbox_inches="tight")
plt.show()
print("Saved: matplotlib_example.png")
````

Expected output:
```text
Saved: matplotlib_example.png
```
The image is saved in the notebook's current working directory.

## 14. Matplotlib with Pandas
Pandas' `DataFrame.plot()` uses Matplotlib by default and is useful after loading and cleaning data.

````python
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Revenue": [1200, 1500, 1350, 1800],
    "Costs": [800, 900, 950, 1100]
})

ax = df.plot(x="Month", y=["Revenue", "Costs"], kind="line", marker="o")
ax.set_title("Revenue vs Costs")
ax.set_ylabel("Amount")
ax.grid(True, alpha=0.25)
plt.show()
print(df)
````

Expected result: two lines for Revenue and Costs, then the four-row DataFrame. The returned Axes object can be customized.

## 15. Choosing the right chart
| Question | Chart | Example |
|---|---|---|
| How does a value change over time? | Line | Monthly sales |
| Which category is larger? | Bar | Members per department |
| Are two numeric variables associated? | Scatter | Study hours vs score |
| How are values distributed? | Histogram | Exam scores |
| What are parts of one total? | Pie, used carefully | Survey shares |
| How do views compare? | Subplots | Revenue and costs |

Avoid 3D effects and decorative clutter. Explain unusual scales and make sure the visualization supports the conclusion.

## 19. Troubleshooting
- `ModuleNotFoundError`: install Matplotlib in the active environment with `python -m pip install matplotlib`.
- No plot in a script: add `plt.show()`.
- Overlapping labels: try `fig.tight_layout()` or increase figure size.
- Missing labels in saved image: save before showing and use `bbox_inches="tight"`.
- Wrong environment: install packages into the same Python environment used by the notebook.
