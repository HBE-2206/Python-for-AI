## 2. What is Matplotlib?
Matplotlib is a Python plotting library used to create static, animated, and interactive visualizations. Its `matplotlib.pyplot` interface offers familiar plotting commands. It is useful during exploratory data analysis (EDA), reporting, and understanding model results.

**Important:** a chart can reveal patterns, but it does not by itself prove causation or guarantee that a model is correct.

## 4. The basic plotting workflow
Most charts follow this pattern:
1. Prepare the data.
2. Call a plotting function.
3. Add a title and labels.
4. Optionally add a legend or grid.
5. Display the figure with `plt.show()`.

`plt.show()` asks Matplotlib to display the current figure. In notebooks, figures may appear automatically, but including it makes scripts clearer.

````python
x = [1,  2,  3,  4,  5]
y = [2,  4,  3,  7,  6]

plt.plot(x, y)
plt.title("A Simple Plot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.show()
````

**Expected result:** a line chart connecting the five points. The horizontal axis is labeled “X values” and the vertical axis “Y values”.

## 5. Line chart — trends over an ordered sequence
Use a line chart when the order matters, such as values over time. Points are connected to make increases, decreases, and changes easier to see.

````python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
sales = [12, 15, 13, 18, 22, 21]

plt.plot(months, sales, marker="o", linewidth=2)
plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales (units)")
plt.grid(True, alpha=0.3)
plt.show()
````

**Expected result:** a line that rises overall, dips in March, rises through May, then dips slightly in June. The grid helps estimate values.

### Multiple lines and legend
Plot multiple series on the same axes when they share the same units and scale. Add a legend so viewers can distinguish them.

````python
months = ["Jan", "Feb", "Mar", "Apr"]
product_a = [10, 14, 13, 18]
product_b = [8, 11, 15, 16]

plt.plot(months, product_a, marker="o", label="Product A")
plt.plot(months, product_b, marker="s", label="Product B")
plt.title("Product Comparison")
plt.xlabel("Month")
plt.ylabel("Sales")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
````

**Expected result:** two labeled lines on one chart. A legend appears identifying Product A and Product B.

## 6. Bar chart — compare categories
Use a bar chart to compare separate categories, such as counts by department or average score by class. Bar charts usually start at zero when bar lengths encode magnitude, because a truncated baseline can exaggerate differences.

````python
departments = ["AI", "Software", "Mechanical", "Electrical"]
members = [12, 18, 10, 14]

plt.bar(departments, members, color="#FF6B35")
plt.title("Team Members by Department")
plt.xlabel("Department")
plt.ylabel("Number of Members")
plt.show()
````

**Expected result:** four vertical bars; Software is highest and Mechanical is lowest.

### Horizontal bar chart
A horizontal chart is often easier to read when category names are long.

````python
courses = ["Python Basics", "Data Analysis", "Machine Learning"]
hours = [10, 16, 24]

plt.barh(courses, hours, color="#FFDC52")
plt.title("Study Hours by Course")
plt.xlabel("Hours")
plt.ylabel("Course")
plt.show()
````

**Expected result:** three horizontal bars, with Machine Learning the longest.

## 7. Scatter plot — relationship between two numeric variables
A scatter plot displays individual pairs of numeric values. It can help reveal clusters, trends, and unusual observations. Correlation in a scatter plot does not establish causation.

````python
study_hours = [1, 2, 2.5, 3, 4, 5, 6, 7]
exam_score = [48, 52, 58, 61, 68, 73, 82, 88]

plt.scatter(study_hours, exam_score, color="#2878B5", s=70)
plt.title("Study Hours vs Exam Score")
plt.xlabel("Study Hours")
plt.ylabel("Exam Score")
plt.grid(True, alpha=0.3)
plt.show()
````

**Expected result:** points generally rise from left to right, suggesting a positive association in this example.

## 8. Histogram — distribution of numeric data
A histogram groups numeric values into intervals called bins and shows how many values fall in each interval. It helps inspect spread, skew, and common ranges. The number and width of bins affect the appearance.

````python
rng = np.random.default_rng(42)
scores = rng.normal(loc=70, scale=10, size=300)

plt.hist(scores, bins=12, edgecolor="white", color="#2878B5")
plt.title("Distribution of Exam Scores")
plt.xlabel("Score")
plt.ylabel("Frequency")
plt.show()
````

**Expected result:** a roughly bell-shaped distribution centered near 70. Since random data is used, the exact bar heights come from the seeded sample.

## 9. Pie chart — parts of a whole
A pie chart shows proportions of a whole. Use it only when categories are mutually exclusive and the values represent parts of the same total. With many categories, a bar chart is usually easier to compare.

````python
labels = ["Python", "SQL", "Excel", "Other"]
values = [40, 25, 20, 15]

plt.pie(values, labels=labels, autopct="%1.1f%%", startangle=90)
plt.title("Tools Used in a Survey")
plt.axis("equal")
plt.show()
````

**Expected result:** four slices labeled with categories and percentages adding to 100%.

## 10. Titles, labels, colors, markers, and line styles
Clear labels explain what the viewer is seeing. Units belong in axis labels when relevant. Choose colors for readability, not decoration alone.

````python
x = [1, 2, 3, 4, 5]
y = [3, 5, 4, 8, 7]

plt.plot(
    x, y,
    color="#FF6B35",
    linestyle="--",
    marker="o",
    markersize=7,
    linewidth=2,
    label="Measured values"
)
plt.title("Customized Line Chart", fontsize=14)
plt.xlabel("Experiment number")
plt.ylabel("Measurement (units)")
plt.legend()
plt.grid(True, alpha=0.25)
plt.show()
````

**Expected result:** an orange dashed line with circular markers, a title, axis labels, legend, and light grid.

## 11. Figure size and subplots
A figure is the overall canvas; an axes is an individual plotting area. Subplots let you compare several views in one figure. Use `fig, axes = plt.subplots(...)` to get the figure and axes objects.

````python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))

axes[0].plot([1, 2, 3, 4], [2, 4, 3, 5], marker="o")
axes[0].set_title("Line Chart")
axes[0].set_xlabel("Step")
axes[0].set_ylabel("Value")

axes[1].bar(["A", "B", "C"], [5, 3, 7])
axes[1].set_title("Bar Chart")
axes[1].set_ylabel("Count")

fig.suptitle("Two Views of Example Data")
fig.tight_layout()
plt.show()
````

**Expected result:** one figure containing a line chart on the left and a bar chart on the right. `tight_layout()` helps prevent labels from overlapping.

## 12. Add annotations
Annotations call attention to a meaningful point. Use them sparingly so they do not clutter the chart.

````python
days = [1, 2, 3, 4, 5]
temperature = [20, 22, 21, 27, 25]

plt.plot(days, temperature, marker="o")
plt.annotate(
    "Highest value",
    xy=(4, 27),
    xytext=(2.8, 28),
    arrowprops={"arrowstyle": "->"}
)
plt.title("Daily Temperature")
plt.xlabel("Day")
plt.ylabel("Temperature (°C)")
plt.show()
````

**Expected result:** a line chart with an arrow pointing to the highest plotted value, 27°C on day 4.

## 13. Save a chart to a file
Call `savefig()` before `show()` for predictable behavior across environments. Use a higher DPI for raster images intended for print; `bbox_inches="tight"` helps include labels.

````python
plt.figure(figsize=(7, 4))
plt.plot([1, 2, 3, 4], [4, 6, 5, 9], marker="o")
plt.title("Saved Example")
plt.xlabel("X")
plt.ylabel("Y")
plt.savefig("matplotlib_example.png", dpi=200, bbox_inches="tight")
plt.show()

print("Saved: matplotlib_example.png")
````

```text
Saved: matplotlib_example.png
```

## 14. Matplotlib with Pandas
Pandas can produce quick charts through its plotting interface, which uses Matplotlib by default. This is convenient after cleaning and grouping tabular data.

````python
df = pd.DataFrame({
    "Month": ["Jan", "Feb", "Mar", "Apr"],
    "Revenue": [1200, 1500, 1350, 1800],
    "Costs": [800, 900, 950, 1100]
})

ax = df.plot(x="Month", y=["Revenue", "Costs"], kind="line", marker="o")
ax.set_title("Revenue vs Costs")
ax.set_ylabel("Amount")
ax.grid(True, alpha=0.25)
plt.show()

print(df)
````

**Expected result:** a two-line chart for Revenue and Costs, followed by the four-row DataFrame. Pandas' `plot()` returns an Axes object, which can be customized using methods such as `set_title()` and `set_ylabel()`.

## 15. Choosing the right chart
| Question | Useful chart | Example |
|---|---|---|
| How does a value change over time? | Line | Monthly sales |
| Which category is larger? | Bar | Members by department |
| Are two numeric variables associated? | Scatter | Study hours vs score |
| How are numeric values distributed? | Histogram | Exam scores |
| What are the parts of one total? | Pie, used carefully | Survey shares |
| How do multiple views compare? | Subplots | Revenue and costs |

Avoid 3D effects and unnecessary decoration. Include units, explain unusual scales, and ensure the visualization supports the conclusion you describe.

## 16. Matplotlib in an AI workflow
A common workflow is:
1. **Pandas** loads and cleans a dataset.
2. **NumPy** supports numeric calculations.
3. **Matplotlib** visualizes distributions, trends, and relationships.
4. **scikit-learn** trains a model.
5. Matplotlib visualizes evaluation results, such as actual vs predicted values or a confusion matrix.

Visualization helps with exploration and communication, but data leakage, biased sampling, and poor evaluation cannot be fixed by a chart alone.

## 19. Troubleshooting
- **`ModuleNotFoundError: No module named 'matplotlib'`:** install Matplotlib in the active environment using `python -m pip install matplotlib`.
- **Plot does not appear in a script:** add `plt.show()`.
- **Labels overlap:** try `fig.tight_layout()` or adjust figure size.
- **Saved image is blank or missing labels:** call `savefig()` before `show()` and use `bbox_inches="tight"`.
- **Wrong Python environment:** run installation in the same environment/kernel used by Jupyter.
