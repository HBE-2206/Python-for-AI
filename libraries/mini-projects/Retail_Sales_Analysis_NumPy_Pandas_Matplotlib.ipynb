## 1. Import the libraries

NumPy supports numerical operations, Pandas organizes the data, and Matplotlib creates visualizations.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use('default')
```

## 2. Create the dataset

Each row represents one product category in one month. Revenue is recorded in fictional currency units.

```python
data = {
    'month': ['Jan', 'Jan', 'Feb', 'Feb', 'Mar', 'Mar',
              'Apr', 'Apr', 'May', 'May', 'Jun', 'Jun'],
    'category': ['Accessories', 'Devices'] * 6,
    'units_sold': [120, 45, 135, 50, 150, 55, 160, 62, 175, 68, 190, 72],
    'unit_price': [15, 120, 15, 120, 15, 120, 15, 120, 15, 120, 15, 120]
}

df = pd.DataFrame(data)
df['revenue'] = df['units_sold'] * df['unit_price']
df.head()
```

**What to notice:** `revenue` is calculated from units sold multiplied by unit price. Pandas creates the new column for every row without a manual loop.

## 3. Inspect and validate the data

Check the dimensions, data types, missing values, and duplicate rows before analyzing the dataset.

```python
print('Rows and columns:', df.shape)
print('\nData types:')
print(df.dtypes)
print('\nMissing values per column:')
print(df.isna().sum())
print('\nDuplicate rows:', df.duplicated().sum())
```

**Interpretation:** The dataset should contain 12 rows and 5 columns. This teaching dataset is designed to have no missing values and no duplicate rows. In a real dataset, investigate unexpected missing values and duplicates before proceeding.

## 4. Calculate summary metrics with NumPy

Use NumPy to calculate total revenue, average revenue per row, and the spread of row-level revenue.

```python
revenue_values = df['revenue'].to_numpy()

total_revenue = np.sum(revenue_values)
average_row_revenue = np.mean(revenue_values)
revenue_std = np.std(revenue_values)

print(f'Total revenue: {total_revenue:,.0f}')
print(f'Average revenue per row: {average_row_revenue:,.2f}')
print(f'Revenue standard deviation (population): {revenue_std:,.2f}')
```

**Important:** the average above is the average revenue per category-month row, not the average order value. Those metrics mean different things.

## 5. Compare revenue by category

Group the data by category and calculate total revenue and units sold.

```python
category_summary = (
    df.groupby('category', as_index=False)
      .agg(total_revenue=('revenue', 'sum'),
           total_units=('units_sold', 'sum'))
      .sort_values('total_revenue', ascending=False)
)

category_summary
```

Now visualize the category comparison. Bar charts work well when comparing totals across a small number of categories.

```python
plt.figure(figsize=(8, 5))
plt.bar(category_summary['category'], category_summary['total_revenue'])
plt.title('Total Revenue by Product Category')
plt.xlabel('Category')
plt.ylabel('Revenue (fictional units)')
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

**Question:** Which category generates more revenue? Consider both unit price and units sold when explaining the result.

## 6. Analyze the monthly trend

Aggregate revenue by month. The month order is explicitly defined so the chart follows the calendar rather than alphabetical order.

```python
month_order = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
monthly_summary = (
    df.groupby('month')['revenue'].sum()
      .reindex(month_order)
      .rename('total_revenue')
      .reset_index()
)

monthly_summary
```

```python
plt.figure(figsize=(9, 5))
plt.plot(monthly_summary['month'], monthly_summary['total_revenue'],
         marker='o', linewidth=2)
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Revenue (fictional units)')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

**Interpretation prompt:** Is revenue increasing, decreasing, or fluctuating across the six months? Describe the pattern without claiming that the same trend must continue in the future.

## 7. Compare category trends over time

A pivot table makes it easy to compare the categories month by month.

```python
category_month = df.pivot(index='month', columns='category', values='revenue')
category_month = category_month.reindex(month_order)
category_month
```

```python
ax = category_month.plot(figsize=(9, 5), marker='o')
ax.set_title('Monthly Revenue by Category')
ax.set_xlabel('Month')
ax.set_ylabel('Revenue (fictional units)')
ax.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
