## 1. Why NumPy?

Python lists are flexible, but NumPy arrays are designed for efficient numerical computation. Arrays commonly contain elements of one data type and support vectorized operations without writing an explicit Python loop.

````python
python_list = [1, 2, 3, 4]
array = np.array([1, 2, 3, 4])
print('List doubled:', [x * 2 for x in python_list])
print('Array doubled:', array * 2)
print('Array type:', type(array))
````

**Expected output**
```text
List doubled: [2, 4, 6, 8]
Array doubled: [2 4 6 8]
Array type: <class 'numpy.ndarray'>
```

Note: multiplying a list by 2 repeats the list; multiplying a NumPy numeric array by 2 multiplies each element.

## 2. Creating arrays

Common constructors include `np.array`, `np.zeros`, `np.ones`, `np.full`, `np.arange`, and `np.linspace`. `arange` uses a step; `linspace` creates a specified number of evenly spaced values.

````python
a = np.array([10, 20, 30])
zeros = np.zeros((2, 3))
ones = np.ones(4, dtype=int)
filled = np.full((2, 2), 7)
sequence = np.arange(0, 10, 2)
points = np.linspace(0, 1, 5)
print('a:', a)
print('zeros shape:', zeros.shape)
print('ones:', ones)
print('filled:\n', filled)
print('arange:', sequence)
print('linspace:', points)
````

**Expected output**
```text
a: [10 20 30]
zeros shape: (2, 3)
ones: [1 1 1 1]
filled:
 [[7 7]
 [7 7]]
arange: [0 2 4 6 8]
linspace: [0.   0.25 0.5  0.75 1.  ]
```

## 3. Dimensions, shape, size and dtype

- `ndim`: number of dimensions.
- `shape`: length along each dimension.
- `size`: total number of elements.
- `dtype`: element data type.

A 2-D array is often interpreted as rows and columns.

````python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print('matrix:\n', matrix)
print('ndim:', matrix.ndim)
print('shape:', matrix.shape)
print('size:', matrix.size)
print('dtype:', matrix.dtype)
````

**Expected output**
```text
matrix:
 [[1 2 3]
 [4 5 6]]
ndim: 2
shape: (2, 3)
size: 6
dtype: int64  # may differ by platform
```

## 4. Data types and conversion

NumPy uses explicit numeric types such as `int32`, `int64`, and `float64`. A conversion may lose information, such as converting a decimal to an integer.

````python
values = np.array([1, 2, 3], dtype=np.float64)
print(values, values.dtype)
converted = np.array([1.9, 2.7, 3.2]).astype(int)
print(converted)
````

**Expected output**
```text
[1. 2. 3.] float64
[1 2 3]
```

`astype(int)` truncates the fractional part here; it does not round to the nearest integer.

## 5. Indexing and slicing

Indexing starts at zero. For 2-D arrays use `[row, column]`. Slices use a stop index that is excluded. Basic slices usually return a view, so editing the slice can affect the original array.

````python
x = np.array([10, 20, 30, 40, 50])
print('first:', x[0])
print('last:', x[-1])
print('slice:', x[1:4])
mat = np.array([[10, 20, 30], [40, 50, 60]])
print('row 2:', mat[1])
print('row 1, col 3:', mat[0, 2])
print('first column:', mat[:, 0])
````

**Expected output**
```text
first: 10
last: 50
slice: [20 30 40]
row 2: [40 50 60]
row 1, col 3: 30
first column: [10 40]
```

## 6. Vectorized arithmetic and universal functions

Operations between compatible arrays are applied elementwise. Useful functions include `np.sqrt`, `np.abs`, `np.round`, `np.exp`, and `np.log`. Prefer vectorized expressions for clear numerical code.

````python
x = np.array([1., 4., 9.])
y = np.array([10., 20., 30.])
print('x + y:', x + y)
print('x * 2:', x * 2)
print('square roots:', np.sqrt(x))
print('rounded:', np.round(y / 7, 2))
````

**Expected output**
```text
x + y: [11. 24. 39.]
x * 2: [2. 8. 18.]
square roots: [1. 2. 3.]
rounded: [1.43 2.86 4.29]
```

## 7. Aggregations and axis

`sum`, `mean`, `median`, `min`, `max`, and `std` summarize data. For a 2-D array, `axis=0` reduces down rows (one result per column); `axis=1` reduces across columns (one result per row).

````python
scores = np.array([[80, 90, 70], [60, 75, 85]])
print('overall mean:', scores.mean())
print('column means:', scores.mean(axis=0))
print('row means:', scores.mean(axis=1))
print('max score:', scores.max())
print('sum by column:', scores.sum(axis=0))
````

**Expected output**
```text
overall mean: 76.66666666666667
column means: [70.  82.5 77.5]
row means: [80.         73.33333333]
max score: 90
sum by column: [140 165 155]
```

## 8. Boolean masks and filtering

Comparisons produce Boolean arrays. Use them as masks to select matching elements. Combine multiple conditions with `&` and `|`, and put each comparison in parentheses; do not use Python's `and`/`or` on whole arrays.

````python
ages = np.array([17, 22, 35, 16, 29])
print('adults:', ages >= 18)
print('adult ages:', ages[ages >= 18])
print('ages 18–30:', ages[(ages >= 18) & (ages <= 30)])
````

**Expected output**
```text
adults: [False  True  True False  True]
adult ages: [22 35 29]
ages 18–30: [22 29]
```

## 9. Reshape, transpose and concatenate

`reshape` changes dimensions without changing the element count. `T` transposes rows and columns. `concatenate` joins arrays along an existing axis; shapes must match on the other axes.

````python
x = np.arange(1, 7)
print('reshape:\n', x.reshape(2, 3))
print('transpose:\n', x.reshape(2, 3).T)
a = np.array([1, 2])
b = np.array([3, 4])
print('joined:', np.concatenate([a, b]))
````

**Expected output**
```text
reshape:
 [[1 2 3]
 [4 5 6]]
transpose:
 [[1 4]
 [2 5]
 [3 6]]
joined: [1 2 3 4]
```

## 10. Broadcasting

Broadcasting lets NumPy operate on arrays with compatible shapes. Dimensions are compatible when they are equal or one of them is 1. This example adds a 1-D vector to every row of a 2-D matrix.

````python
matrix = np.array([[1, 2, 3], [4, 5, 6]])
offset = np.array([10, 100, 1000])
print(matrix + offset)
````

**Expected output**
```text
[[  11  102 1003]
 [  14  105 1006]]
```

Broadcasting does not mean arbitrary shapes work; incompatible shapes raise a `ValueError`.

## 11. Sorting, unique values and counts

`np.sort` returns a sorted copy. `np.unique` returns unique values; with `return_counts=True`, it also returns each value's frequency.

````python
values = np.array([3, 1, 3, 2, 1, 3])
print('sorted:', np.sort(values))
unique, counts = np.unique(values, return_counts=True)
print('unique:', unique)
print('counts:', counts)
````

**Expected output**
```text
sorted: [1 1 2 3 3 3]
unique: [1 2 3]
counts: [2 1 3]
```

## 12. Random numbers and reproducibility

Use `np.random.default_rng(seed)` for modern random-number generation. A fixed seed makes the generated sequence reproducible in the same NumPy environment. Random outputs are examples, not guarantees about real data.

````python
rng = np.random.default_rng(42)
print('integers:', rng.integers(0, 10, size=5))
print('normal shape:', rng.normal(loc=0, scale=1, size=(2, 3)).shape)
````

**Expected output**
```text
integers: [0 7 6 4 4]
normal shape: (2, 3)
```

The exact seeded sequence is stable for this example under the current Generator implementation; the shape line simply confirms a 2-by-3 result.

## 13. Linear algebra basics

For matrix multiplication use `@` or `np.matmul`; `*` is elementwise multiplication. `np.linalg` provides common linear algebra routines.

````python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])
print('elementwise:\n', A * B)
print('matrix product:\n', A @ B)
print('determinant:', round(np.linalg.det(A), 2))
````

**Expected output**
```text
elementwise:
 [[ 5 12]
 [21 32]]
matrix product:
 [[19 22]
 [43 50]]
determinant: -2.0
```

## 14. Mini-project: simple feature normalization

In machine learning, features may have very different scales. This example standardizes each column using its mean and standard deviation. It is for learning; real pipelines should fit normalization parameters on training data only to avoid data leakage.

````python
# Rows are samples; columns are features
X = np.array([[10., 100.], [20., 150.], [30., 200.]])
mean = X.mean(axis=0)
std = X.std(axis=0)
# Guard against zero-variance columns in reusable code
std_safe = np.where(std == 0, 1, std)
X_scaled = (X - mean) / std_safe
print('mean:', mean)
print('std:', std)
print('scaled:\n', np.round(X_scaled, 2))
print('scaled column means:', np.round(X_scaled.mean(axis=0), 8))
````

**Expected output**
```text
mean: [ 20. 150.]
std: [ 8.16496581 40.82482905]
scaled:
 [[-1.22 -1.22]
 [ 0.    0.  ]
 [ 1.22  1.22]]
scaled column means: [0. 0.]
```

## 15. Common mistakes and good practices

- Use `array.shape` and `array.dtype` to inspect data before calculations.
- Distinguish elementwise `*` from matrix multiplication `@`.
- Match dimensions before reshaping or combining arrays.
- Use parentheses with multiple Boolean conditions.
- Be aware that integer arrays cannot preserve fractional values when assigned integer results.
- Avoid relying on a particular random output unless you set and document a seed.
- For data analysis, understand missing values and invalid values before aggregating.
