## 1. Functions

A function is a reusable block of code. Parameters are names in the definition; arguments are values passed to a call. `return` sends a result back to the caller, while `print()` only displays it.

````python
def add(a, b):
    """Return the sum of two numbers."""
    return a + b

result = add(4, 6)
print(result)
````

**Expected output**
```text
10
```

## 2. Default and keyword arguments

A default argument is used when the caller omits that parameter. Keyword arguments make calls more explicit.

````python
def greet(name, message="Welcome"):
    return f"{message}, {name}!"

print(greet("Hazem"))
print(greet("Mona", message="Hello"))
````

**Expected output**
```text
Welcome, Hazem!
Hello, Mona!
```

**Practice:** Create a function that calculates a price after applying a discount percentage.

## 3. Scope and docstrings

Local variables are defined inside a function. Prefer passing needed values as parameters instead of relying unnecessarily on global state. A docstring documents a function's purpose.

````python
def square(number):
    """Return the square of a number."""
    return number ** 2

print(square(5))
print(square.__doc__)
````

**Expected output**
```text
25
Return the square of a number.
```


## 4. List comprehensions and built-ins

Comprehensions transform or filter iterables concisely. Use them for simple operations; prefer a normal loop when the logic becomes hard to read. `enumerate()` provides an index and item; `zip()` pairs items from iterables.

````python
scores = [0.92, 0.45, 0.87, 0.31]
high_scores = [score for score in scores if score >= 0.80]
print(high_scores)

features = ["age", "fare"]
for index, feature in enumerate(features):
    print(index, feature)
````

**Expected output**
```text
[0.92, 0.87]
0 age
1 fare
```


## 5. Errors and exceptions

Syntax errors prevent parsing. Exceptions occur during execution. Logic errors produce an incorrect result even when the program runs. Handle specific expected exceptions rather than swallowing all errors.

````python
try:
    number = int("42")
except ValueError:
    print("Invalid number")
else:
    print(number)
finally:
    print("Finished")
````

**Expected output**
```text
42
Finished
```

## 6. File handling

Use `with open(...)` so the file is closed automatically. `w` creates or overwrites, `a` appends, and `r` reads. Use an explicit encoding for text.

````python
with open("notes.txt", "w", encoding="utf-8") as file:
    file.write("Learning Python\\n")

with open("notes.txt", "r", encoding="utf-8") as file:
    print(file.read().strip())
````

**Expected output**
```text
Learning Python
```


## 7. JSON

JSON is a text format for structured data. `json.dumps()` serializes a Python object to a JSON string; `json.loads()` parses a JSON string into a Python object.

````python
import json

student = {"name": "Hazem", "score": 92}
text = json.dumps(student)
restored = json.loads(text)
print(restored["name"], restored["score"])
````

**Expected output**
```text
Hazem 92
```


## 8. Object-oriented programming

A class defines a type; an object is an instance of that class. Attributes hold data and methods define behavior. `self` refers to the current instance.

````python
class Student:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def introduce(self):
        return f"My name is {self.name}; I am {self.age}."

student = Student("Harry", 23)
print(student.introduce())
````

**Expected output**
```text
My name is Harry; I am 23.
```


## 9. Inheritance and polymorphism

Inheritance allows a child class to reuse or extend a parent class. Polymorphism allows different objects to implement the same interface in different ways.

````python
class Animal:
    def speak(self):
        return "Some sound"

class Dog(Animal):
    def speak(self):
        return "Woof!"

print(Dog().speak())
````

**Expected output**
```text
Woof!
```


## 10. Modules and packages

A module is a Python file. A package organizes related modules. Imports let code reuse functions, classes and constants defined elsewhere.

````python
import math
from pathlib import Path

print(math.sqrt(25))
print(Path("data.txt").suffix)
````

**Expected output**
```text
5.0
.txt
```


## 11. Virtual environments

A virtual environment isolates installed dependencies for a project. Create and activate it using the command for your operating system and shell. These commands are examples, not universal for every terminal.

````python
# Create an environment
python -m venv .venv

# Windows PowerShell activation
.venv\\Scripts\\Activate.ps1

# macOS/Linux activation
source .venv/bin/activate

# Install a package in the active environment
python -m pip install numpy
````

The activation command differs by operating system and shell. Run only the command appropriate to your environment.

## 12. Testing with assert

`assert` is useful for quick checks during learning. Assertions can be disabled when Python runs with optimization (`-O`), so do not use them for essential runtime validation. Larger projects should use a test framework such as `unittest` or `pytest`.

````python
def add(a, b):
    return a + b

assert add(2, 3) == 5
assert add(-1, 1) == 0
print("Basic tests passed")
````

**Expected output**
```text
Basic tests passed
```
