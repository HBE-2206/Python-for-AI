```python
import torch
batch,seq,d_model,heads=2,8,32,4; dim=d_model//heads
q=torch.randn(batch,heads,seq,dim); k=torch.randn_like(q); v=torch.randn_like(q)
scores=(q@k.transpose(-2,-1))/(dim**.5); weights=torch.softmax(scores,dim=-1); context=weights@v
print('attention',weights.shape,'context',context.shape)
# Tensor-shape demonstration only; not a complete Transformer.
```
