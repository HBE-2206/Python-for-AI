```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report,roc_auc_score
X,y=load_breast_cancer(return_X_y=True); Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,stratify=y,random_state=42)
m=make_pipeline(StandardScaler(),LogisticRegression(max_iter=2000)).fit(Xtr,ytr); p=m.predict(Xte)
print(classification_report(yte,p)); print('ROC-AUC',roc_auc_score(yte,m.predict_proba(Xte)[:,1]))
```
