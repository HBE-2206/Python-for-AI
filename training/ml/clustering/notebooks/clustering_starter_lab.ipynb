```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans,DBSCAN
from sklearn.metrics import silhouette_score
X,_=make_moons(n_samples=500,noise=.07,random_state=42); X=StandardScaler().fit_transform(X)
fig,axs=plt.subplots(1,2,figsize=(10,4))
for ax,(n,m) in zip(axs,{'K-Means':KMeans(n_clusters=2,n_init=10,random_state=42),'DBSCAN':DBSCAN(eps=.25,min_samples=5)}.items()):
 labels=m.fit_predict(X); ax.scatter(X[:,0],X[:,1],c=labels,s=12); ax.set_title(n); mask=labels!=-1
 if len(set(labels[mask]))>1: print(n,silhouette_score(X[mask],labels[mask]))
plt.show()
```
