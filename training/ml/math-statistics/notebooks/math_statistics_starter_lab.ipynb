```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
rng=np.random.default_rng(42)
x=rng.normal(10,2,500); y=3*x+rng.normal(0,3,500)
df=pd.DataFrame({'feature':x,'target':y})
print(df.describe()); print(df.corr())
df.hist(figsize=(8,3)); plt.tight_layout(); plt.show()
A=np.array([[1,2],[3,4]]); print(A@np.array([2,1]))
```
