```python
from sklearn.datasets import fetch_california_housing
from sklearn.model_selection import train_test_split
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression,Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
X,y=fetch_california_housing(return_X_y=True,as_frame=True); Xtr,Xte,ytr,yte=train_test_split(X,y,test_size=.2,random_state=42)
models={'Linear':make_pipeline(StandardScaler(),LinearRegression()),'Ridge':make_pipeline(StandardScaler(),Ridge()),'Forest':RandomForestRegressor(n_estimators=100,random_state=42,n_jobs=-1)}
for n,m in models.items():
 m.fit(Xtr,ytr); p=m.predict(Xte); print(n,mean_absolute_error(yte,p),mean_squared_error(yte,p)**.5,r2_score(yte,p))
```
