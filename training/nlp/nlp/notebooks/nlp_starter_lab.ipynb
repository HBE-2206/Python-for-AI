```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import make_pipeline
texts=['great product','excellent quality','terrible service','bad experience','happy purchase','not worth money']; y=[1,1,0,0,1,0]
m=make_pipeline(TfidfVectorizer(ngram_range=(1,2)),LogisticRegression()).fit(texts,y)
print(m.predict(['great quality','terrible product']))
```
