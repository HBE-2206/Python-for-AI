```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid", context="notebook")
RANDOM_STATE = 42
```

## 1. Create a reproducible sample dataset
We use synthetic data so the notebook runs without downloading an external dataset. It is for learning visualization patterns, not for drawing real-world conclusions.

```python
rng = np.random.default_rng(RANDOM_STATE)
n = 240
df = pd.DataFrame({
    "study_hours": rng.uniform(0.5, 8.0, n).round(1),
    "attendance": rng.uniform(55, 100, n).round(1),
    "group": rng.choice(["A", "B", "C"], size=n, p=[0.35, 0.35, 0.30]),
})
group_effect = df["group"].map({"A": 2, "B": 0, "C": -2})
df["score"] = (45 + 4.2 * df["study_hours"] + 0.22 * df["attendance"] + group_effect + rng.normal(0, 7, n)).clip(0, 100).round(1)
df.head()
```

## 2. Inspect the data before plotting
Check types, missing values, and summary statistics so the chart is based on an understanding of the dataset.

```python
display(df.info())
display(df.describe(numeric_only=True))
display(df.isna().sum())
```

## 3. Histogram — inspect a distribution
Use a histogram to understand how a numeric variable is distributed.

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.hist(df["score"], bins=15, edgecolor="black")
ax.set(title="Distribution of Student Scores", xlabel="Score", ylabel="Number of observations")
fig.tight_layout()
plt.show()
```

## 4. Scatter plot — explore relationships
A scatter plot can show association between two numeric variables. Association alone does not prove causation.

```python
fig, ax = plt.subplots(figsize=(8, 4.5))
for group, subset in df.groupby("group"):
    ax.scatter(subset["study_hours"], subset["score"], alpha=0.65, label=f"Group {group}")
ax.set(title="Study Hours vs. Score", xlabel="Study hours", ylabel="Score")
ax.legend(title="Group")
fig.tight_layout()
plt.show()
```

## 5. Bar chart — compare categories
A bar chart compares a summary statistic across categories. State which statistic is being shown.

```python
mean_scores = df.groupby("group", as_index=False)["score"].mean().sort_values("group")
fig, ax = plt.subplots(figsize=(7, 4.5))
ax.bar(mean_scores["group"], mean_scores["score"])
ax.set(title="Mean Score by Group", xlabel="Group", ylabel="Mean score", ylim=(0, 100))
fig.tight_layout()
plt.show()
mean_scores
```

## 6. Box plot — compare spread and potential outliers
Box plots summarize the median, quartiles, and potential outliers; they do not show every detail of a distribution.

```python
fig, ax = plt.subplots(figsize=(7, 4.5))
sns.boxplot(data=df, x="group", y="score", ax=ax)
ax.set(title="Score Spread by Group", xlabel="Group", ylabel="Score")
fig.tight_layout()
plt.show()
```

## 7. Correlation heatmap
Correlation summarizes linear association between numeric variables. It does not establish causation and can be affected by confounding variables.

```python
numeric_corr = df[["study_hours", "attendance", "score"]].corr()
fig, ax = plt.subplots(figsize=(6, 4.5))
sns.heatmap(numeric_corr, annot=True, fmt=".2f", cmap="vlag", center=0, vmin=-1, vmax=1, ax=ax)
ax.set_title("Correlation Between Numeric Variables")
fig.tight_layout()
plt.show()
```
